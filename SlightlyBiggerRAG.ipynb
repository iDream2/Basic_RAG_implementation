```python
from dotenv import load_dotenv
load_dotenv()
client = OpenAI(api_key = os.getenv("OPENAI_API_KEY"))

pdf  = fitz.open("Documents/RAG_Document_1_Python.pdf") 
pdf2 = fitz.open("Documents/RAG_Document_2_Docker.pdf")
pdf3 = fitz.open("Documents/RAG_Document_3_Cloud.pdf")
```

```python
doc1 = ""
for page in pdf:
     doc1 += page.get_text() 

print("The text in pdf 1 is - \n" + doc1)
```

```python
doc2 = ""
for page in pdf2:
     doc2 += page.get_text() 

print("The text in pdf 2 is - \n" + doc2)
```

```python

doc3 = ""
for page in pdf3:
     doc3 += page.get_text() 

print("The text in pdf 3 is - \n" + doc3)
```

```python
documents = [doc1, doc2, doc3]
```

```python
#Now the idea is to split them up and convert them into embeddings

#Step-1 : Function for returning a single chunk of 50 words

def chunkDocuments(text, chunk_size = 50):
    words = text.split()

    chunks = []

    for i in range( 0, len(words), chunk_size ):
          chunk = "".join(words[i:i + chunk_size])
          chunks.append(chunk)

    return chunks 


all_chunks = []
```

```python
for doc in documents:
     chunks  =  chunkDocuments(doc) #So, chunks is a list
     all_chunks.extend(chunks)
```

```python
print("Number of Chunks:", len(all_chunks))

for i, chunk in enumerate(all_chunks):
     print(f"""\n --- Chunk{i+1}""")
     print(chunk[:200])
```

```python

#Step-2 Create embeddings for the chunks - 
embeddings = []

for chunk in all_chunks:
     result = client.embeddings.create(
          model = "text-embedding-3-small", #The model that we use for embedding
          input=chunk
     )

     embeddings.append(result.data[0].embedding)


print("Embedding of Chunk 1 : " , embeddings[0] )
print(f"""For 15 chunks we need to have {len(embeddings)} embeddings""")
# question = "Tell me something about Cloud?"
```

```python
#Now the idea is to convert the question into chunks

question = "Tell me something about Cloud Computing"
questionEmbeddingResponse = client.embeddings.create(
    model = "text-embedding-3-small",
    input = question
)

questionEmbedding = questionEmbeddingResponse.data[0].embedding 
print(questionEmbedding)
#Embed this question - 
```

```python
#Now the idea is to match the question with the embedding of the chunks ~ 
!pip install numpy
```

```python
import numpy as np

def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2)/( np.linalg.norm(vec1) * np.linalg.norm(vec2) )
```

```python
similarityScoreVector = []

for i, chunkEmbeddings in enumerate(embeddings):
    similarityScore = cosine_similarity(questionEmbedding,chunkEmbeddings)
    similarityScoreVector.append((similarityScore,i)) 
```

```python
print(f"""Similarity Score with Chunk {similarityScoreVector[1][1]}:""", similarityScoreVector[1])
```

```python
#Now select the top-k values. We are selecting the top-3 
#Let's sort the vector in a  reverse manner to get the top-3 values which will give us the top-3 values and their chunk number

similarityScoreVector.sort(reverse=True)
print(similarityScoreVector)
```

```python
top3Values = [similarityScoreVector[0], similarityScoreVector[1],similarityScoreVector[2] ]
print(top3Values)
```

```python
top3Chunks = []

for scores,index in top3Values:
    top3Chunks.append(all_chunks[index])
```

```python
context = f"""{top3Chunks[0]} \n {top3Chunks[1]} \n {top3Chunks[2]}"""

prompt = f"""Answer the following question with the following context: 

question : {question}

context : {context}

If the answer cannot be found in the context, say that the provided documents do not contain enough information. ALso fetch some additional information from the internet and put it under Addition Information
"""
```

```python
response = client.responses.create(
    model= "gpt-4o-mini",
    input= prompt
)
print(response.output_text)
```
